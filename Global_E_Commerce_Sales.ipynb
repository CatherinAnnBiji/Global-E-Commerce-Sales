```python
df['Profit_margin%'] = np.where(df['Total_Sales'] != 0,(df['Profit'] / df['Total_Sales']) * 100, 0)
```

```python
bin1 = [-100,5,10,20,100]
label1 = ['low_profit','Average_profit','Medium_profit','High_profit']
df['Profit_remake'] = pd.cut(df['Profit_margin%'], bins=bin1, labels=label1,right=False)
```

```python
df['Discount_category'] = np.where(df['Discount_Percent'] == 0, 'No Discount',np.where(df['Discount_Percent'] <= 10, 'Low',np.where(df['Discount_Percent'] <= 20, 'Medium', 'High')))
```

```python
df['Year']=df['Order_Date'].dt.year
```

```python
df['Year']=df['Year'].astype(str)
```

```python
df['Month']=df['Order_Date'].dt.month_name()
df
```

Sales Analysis

```python
plt.figure(figsize=(14,4))
sns.lineplot(data=Monthly_sales2023, x='Month',y='Total_sales',color='black',marker='o',linewidth=2,linestyle='-',label='2023')
sns.lineplot(data=Monthly_sales2024,x='Month',y='Total_sales',color='teal',marker='o',linewidth=2,label='2024')
sns.lineplot(data=Monthly_sales2025,x='Month',y='Total_sales',color='darkblue',marker='o',linewidth=2,label='2025')
plt.title('Monthly sales comparsion of 2023, 2024 and 2025', fontsize=14, fontweight='bold')
plt.xlabel('Months')
plt.ylabel('Total Sales')
plt.legend(title='Year',loc='upper right')
plt.xticks(rotation=35)
plt.show()
```

Sales fluctuate significantly month-to-month, showing distinct seasonal peaks and troughs across different years
All three years experience a major surge around October, likely driven by Q4 promotional campaigns, pre-holiday shopping, or annual inventory sales.

Align inventory management with seasonal patterns. Boost stock levels ahead of key demand windows (especially September–October and May–June) to avoid stockouts during high-demand months.

Near-Perfect Sales & Profit Correlation ($0.98$): Total Sales and Profit share an extremely strong positive correlation ($0.98$), indicating that driving top-line revenue directly drives bottom-line profitability.

Shipping Cost Correlates with Quantity ($0.77$): Shipping Cost strongly correlates with Quantity ($0.77$), reflecting that bulkier/larger orders cost more to fulfill and ship.   
Negative Impact of Discounts ($-0.14$): Discount Percent shows a negative correlation with Profit ($-0.14$), mathematically confirming that higher discount percentages reduce overall profitability.
